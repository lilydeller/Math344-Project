```python
import matplotlib.pyplot as plt

n_test = 9968 # len(test), from notebook 02

# test RMSE, from notebooks 02 and 03
results = {
# notebook 2: "scores["always 3"] = rmse(np.full(len(test), 3.0), truth)"
"always 3": 1.2516,
# notebook 2: "scores["overall average"] = rmse(np.full(len(test), mu), truth)"
"overall average": 1.1349,
# notebook 2: "scores["user average"] = rmse(user_mean[u], truth)"
"user average": 1.0452,
# notebook 2: "scores["movie rating"] = rmse(item_mean[i], truth)"
"movie average": 1.0280,
# notebook 2: "scores["user + movie effects"] = rmse(pred, truth)"
"user + movie effects": 0.9606,

# notebook 3: Part G and H
"low rank (k = 10)": 0.9725,
"iterative SVD (k = 2)": 0.9431
}
colors = ["tab:green" if "k =" in name else "gray" for name in results]
plt.figure(figsize=(8, 4))
bars = plt.barh(list(results), list(results.values()), color=colors)
plt.bar_label(bars, fmt="%.3f", padding=3)
plt.gca().invert_yaxis()
plt.xlabel("test RMSE (stars, lower is better)")
plt.title(f"Predicting {n_test:,} hidden MovieLens ratings")
plt.xlim(0, 1.6)
plt.tight_layout()
plt.savefig("hero.png", dpi=150)
plt.show()
```
